2.Character Count 
●Count the number of vowels, consonants, digits, spaces, and special characters in a given string. 

```python
string = input("Enter the string to count its length: ")

vowels = "aeiouAEIOU"
vowelCount = 0
digitCount = 0
consonantCount = 0
specialCharCount = 0
spaceCount = 0


for ch in string:
    if ch in vowels:
        vowelCount += 1
    elif ch.isdigit():
        digitCount += 1
    elif ch.isalpha():
        consonantCount += 1
    elif ch == " ":
        spaceCount += 1
    else:
        specialCharCount += 1

print(f"Vowels: {vowelCount}")
print(f"Digits: {digitCount}")
print(f"Consonants: {consonantCount}")
print(f"Special Characters: {specialCharCount}")
print(f"Spaces: {spaceCount}")
```

3.Reverse a String 
Reverse the given string without using built-in reverse functions

```python
def reverseString(string):
    revStr = ""
    i = len(string) - 1
    while i >= 0:
        revStr += string[i]
        i -= 1
    return revStr
string = input("Enter a string to reverse: ")
print(reverseString(string))
```

4. Palindrome Check 
Check whether the entered string is a palindrome

```python
string = input("Enter a string: ")
reverse = string[::-1]

if string == reverse:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

5.Uppercase and Lowercase Count 
●Count the number of uppercase and lowercase letters in a string. 

```python
string = input("Enter a string: ")

lowerCount = 0
upperCount = 0

for ch in string:
    if ch.islower():
        lowerCount += 1
    elif ch.isupper():
        upperCount += 1
print(f"Lowercase letters: {lowerCount}")
print(f"Uppercase letters: {upperCount}")
```
