# Sets

```python
# Q1. Basic Set Operations
# Write a program using the following set:
# A = {10, 20, 30, 40}
# B = {30, 40, 50, 60}
# Perform:
# add() → Add 70 to A
# remove() → Remove 20 from A
# discard() → Discard 100 from A
# pop() → Remove an arbitrary element
# union() → A and B
# intersection() → A and B

A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
A.add(70)
print(f"After adding 70 to A: {A}")
A.remove(70)
print(f"After removing 20 from A: {A}")
A.discard(100)
print(f"After discarding 100 from A: {A}")
A.pop()
print(f"After popping an arbitrary element from A: {A}")
print(f"Union of A and B: {A.union(B)}")
print(f"Intersection of A and B: {A.intersection(B)}")
```

```python
# Q2. Set Comparison Operations
# A = {1, 2, 3, 4, 5}
# B = {3, 4, 5, 6, 7}
# Write code to find:
# difference() → A - B
# difference() → B - A
# symmetric_difference()
# Check issubset()
# Check issuperset()
# Check isdisjoint()

A = {1, 2, 3, 4, 5}
B = {3, 4, 5, 6, 7}
print(f"diff. between A and B: {A.difference(B)}")
print(f"diff. bet^n B and A: {B.difference(A)}")
print(f"symmetric difference between a and B is : {A.symmetric_difference(B)}")
print(f"is a subset of b?: {A.issubset(B)}")
print(f"is a superset of b? : {A.issuperset(B)}")
print(f"is a disjoint b? : {A.isdisjoint(B)}")
```

```python
# Q3. Set Modification
# A = {10, 20, 30, 40}
# Write a program that:
# Adds 50
# Removes 20
# Discards 100
# Creates a copy using copy()
# Clears the original set using clear()

A = {10, 20, 30, 40}
A.add(50)
print(f"After adding 50: {A}")
A.remove(20)
print(f"After removing 20: {A}")
A.discard(100)
print(f"After discarding 100: {A}")
A_copy = A.copy()
print(f"Copy: {A_copy}")
A.clear()
print(f"After clearing A: {A}")
```

# Frozen Set

```python
# Q4. Frozen Set Operations
# A = frozenset({1, 2, 3, 4})
# B = frozenset({3, 4, 5, 6})
# Perform:
# union()
# intersection()
# difference()
# symmetric_difference()

A = frozenset({1, 2, 3, 4})
B = frozenset({3, 4, 5, 6})
print(f"Union of A and B:{A.union(B)}")
print(f"Intersection of A and B: {A.intersection(B)}")
print(f"difference: {A.difference(B)}")
print(f"Symmetric difference: {A.symmetric_difference(B)}")
```

```python
# Frozen Set Relationships
# A = frozenset({1, 2})
# B = frozenset({1, 2, 3, 4})
# C = frozenset({5, 6})
# Check:
# A.issubset(B)
# B.issuperset(A)
# A.isdisjoint(C)
# len(B)
# Whether 3 exists in B

A = frozenset({1, 2})
B = frozenset({1, 2, 3, 4})
C = frozenset({5, 6})
print(f"Is A subset of B?: {A.issubset(B)}")
print(f"Is B superset of A?: {B.issuperset(A)}")
print(f"Is A disjoint with C?: {A.isdisjoint(C)}")
print(f"Length of Frozen Set B: {B.__len__()}")
print(f"Whether 3 exists in B? : {B.__contains__(3)}")
```

```python
# Q6. Frozen Set as Dictionary Key ⭐
# Create:
# A = frozenset({1, 2, 3})
# B = frozenset({4, 5, 6})
# Create a copy of A using copy()
# Check whether A and B are disjoint
# Use A as a dictionary key
# Store "Numbers" as its value
# Access and print the value


A = frozenset({1, 2, 3})
B = frozenset({4, 5, 6})
A_copy = A.copy()
print(A_copy)
print(A.isdisjoint(B))
my_dict = {A: "Numbers"}
print(my_dict[A])
```

# Dictionary

```python
# Q7. Dictionary Access
# Given:
# student = {
#     "name": "Suyash",
#     "age": 21,
#     "marks": 85
# }
# Write code to:
# Print keys()
# Print values()
# Print items()
# Get "name" using get()
# Get "city" using get() without causing an error

student = {
    "name": "Suyash",
    "age": 21,
    "marks": 85
}
print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))
print(student.get("city"))
```

```python
# Q8. Dictionary Modification
# Given:
# student = {
#     "name": "Suyash",
#     "age": 21,
#     "marks": 85
# }
# Perform:
# update() → Change marks to 95
# update() → Add "city": "Kolhapur"
# setdefault() → Add "branch": "CSE"
# pop() → Remove "age"
# popitem() → Remove the last item
# copy() → Create a copy

student = {
    "name": "Suyash",
    "age": 21,
    "marks": 85
}
print(student.update({"marks": 95}))
print(student.update({"city": "Kolhapur"}))
print(student.setdefault("branch", "CSE"))
print(student.pop("age"))
print(student.popitem())
student_copy = student.copy()
print(student_copy)
```

```python
# Complete Dictionary Challenge ⭐
# Write a program that starts with:
# student = {
#     "name": "Suyash",
#     "age": 21,
#     "marks": 85
# }
# Perform all of these in one program:
# keys()
# values()
# items()
# get()
# update()
# setdefault()
# pop()
# popitem()
# copy()
# clear()

student = {
    "name": "Suyash",
    "age": 21,
    "marks": 85
}
print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))
student.update({"marks": 95})
student.setdefault("city", "Kolhapur")
student.pop("age")
student.popitem()
student_copy = student.copy()
print(student_copy)
student.clear()
print(student)
```

# Lambda Function

```python
marks = [95, 82, 68, 55, 35, 91, 73, 45]

result = list(map(lambda x: "Excellent" if x >= 90 else "Very Good" if x >= 75 else "Good" if x >= 60 else "Pass" if x >= 40 else "Fail", marks))

print(result)
```

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
sentence = "Python programming is very interesting"
words = word_tokenize(sentence)
result = list(map(lambda x: "Long" if len(x) >= 7 else "Medium" if len(x) >= 4 else "Short", words))
print(result)
```
